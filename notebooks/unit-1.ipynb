Esta presentación introduce la motivación del curso y el papel de los Transformers dentro del desarrollo reciente de la inteligencia artificial generativa.


## Bienvenida

Este curso estudia los fundamentos de los Transformers desde una perspectiva conceptual, geométrica y computacional. La ruta se concentra en comprender cómo una secuencia se transforma en representaciones internas, cómo opera la atención y cómo esos componentes se integran en un modelo autoregresivo pequeño.

El punto de partida no es usar un sistema ya construido, sino abrir la arquitectura para entender sus piezas principales.


## Por qué Transformers

El estado actual de muchas tecnologías de inteligencia artificial depende de varias capas: infraestructura de cómputo, datos, optimización, estadística, matemáticas y arquitecturas capaces de escalar.

Dentro de esa historia, el Transformer ocupa un lugar central porque hizo operativa la idea de autoatención: cada posición de una secuencia puede comparar su representación con otras posiciones y construir una nueva representación según las relaciones relevantes.

Esa operación permitió modelar dependencias complejas de manera paralela y flexible, dentro de una ventana de contexto, sin depender únicamente de procesar la secuencia paso a paso.


## Alcance

El curso no es una introducción a ChatGPT, prompting, RAG, agentes ni ajuste de modelos grandes. Esos temas pertenecen a otra capa de uso y aplicación.

Aquí el objetivo es entender los mecanismos fundamentales: tokens, embeddings, posición, formas tensoriales, atención, Q, K, V, self-attention, máscara causal, multi-head attention, bloque decoder, mini-GPT, entrenamiento autoregresivo, generación y límites.
